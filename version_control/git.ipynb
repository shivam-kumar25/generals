# Practical Git + GitHub Notes

Git is used by developers to **track code changes, work safely in branches, collaborate through GitHub, review changes through PRs, and recover from mistakes**. Think of Git as your local project history machine, and GitHub as the hosted collaboration platform.

```mermaid
flowchart LR
    A[Working directory<br/>files you edit] -->|git add| B[Staging area<br/>selected changes]
    B -->|git commit| C[Local Git repo<br/>history on laptop]
    C -->|git push| D[GitHub remote<br/>shared repo]
    D -->|git fetch / pull| C
```

Git stores snapshots of your project, not just file versions. Branches are lightweight pointers to commits, which is why branching is fast and cheap.

Start once on your laptop:

```bash
# Ubuntu / Debian
sudo apt update
sudo apt install git

# macOS
brew install git

# Windows
winget install -e --id Git.Git

# Check installation
git --version

# One-time identity setup
git config --global user.name "Your Name"
git config --global user.email "your-email@example.com"

# Good modern defaults
git config --global init.defaultBranch main
git config --global pull.rebase true
git config --global core.editor "code --wait"

# See all global settings
git config --global --list
```

The daily Git loop is small: **status → diff → add → commit → log**.

```bash
# Create a new project repo
mkdir my-project
cd my-project
git init

# Create first file
echo "# my-project" > README.md

# See what changed
git status

# See exact file changes
git diff

# Stage selected file
git add README.md

# Save staged change into history
git commit -m "docs: add README"

# See short history
git log --oneline
```

The staging area is important because you do **not** have to commit everything you edited. You can make clean, small commits.

```bash
# Suppose you edited 3 files but want only 2 in this commit
git add src/app.py tests/test_app.py
git commit -m "feat(app): add greeting endpoint"

# Stage only parts of a file
git add -p app.py

# See staged changes before committing
git diff --staged
```

Good commits are small and explain intent. Use Conventional Commits: `type(scope): subject`. Common types: `feat`, `fix`, `docs`, `test`, `refactor`, `chore`, `ci`.

```bash
git commit -m "feat(api): add user search"
git commit -m "fix(auth): reject expired tokens"
git commit -m "docs(readme): add setup steps"
git commit -m "test(api): cover empty response"
git commit -m "chore(deps): update lockfile"
```

Branches let you work without damaging `main`.

```mermaid
gitGraph
    commit id: "init"
    commit id: "readme"
    branch feature/login
    checkout feature/login
    commit id: "login form"
    commit id: "validation"
    checkout main
    commit id: "hotfix"
    checkout feature/login
    commit id: "tests"
```

```bash
# Create and switch to a branch
git switch -c feature/login

# Work normally
echo "login work" > login.txt
git add login.txt
git commit -m "feat(login): add login draft"

# Go back to main
git switch main

# Merge branch into main
git merge feature/login

# Delete local branch after merge
git branch -d feature/login
```

Use `git switch` for branches and `git restore` for files. Older tutorials use `git checkout`, but modern Git split branch switching and file restoring into clearer commands.

Connect local Git to GitHub:

```bash
# Add GitHub repo as remote
git remote add origin https://github.com/USERNAME/my-project.git

# Rename current branch to main
git branch -M main

# Push first time and set upstream
git push -u origin main

# After upstream is set
git push
git pull
```

Use `fetch` when you want to inspect remote changes before merging.

```bash
# Download remote updates but do not merge
git fetch origin

# See difference between local main and remote main
git log --oneline main..origin/main

# Then merge/rebase when ready
git pull
```

Safe undo commands matter more than advanced commands.

```bash
# Discard unstaged changes in one file
git restore app.py

# Unstage a file but keep changes
git restore --staged app.py

# Fix last commit message
git commit --amend -m "docs: improve README setup"

# Add forgotten file to previous commit
git add missing_file.py
git commit --amend --no-edit

# Undo last commit but keep changes
git reset HEAD~1

# Create a new commit that reverses an old commit
git revert <commit-sha>

# Dangerous: deletes uncommitted work
git reset --hard HEAD~1

# Recovery history if you moved HEAD accidentally
git reflog
```

Safe habit: **prefer `git revert` on shared branches**. Use `reset` only when you understand whether the commits are private or already pushed.

Rebase means “replay my branch commits on top of a newer base.” It keeps history linear, but it rewrites commits, so do not rebase commits that teammates already pulled.

```mermaid
flowchart TB
    subgraph Before
        A1[main: A] --> B1[main: B]
        A1 --> C1[feature: C]
        C1 --> D1[feature: D]
    end

    subgraph After_rebase
        A2[main: A] --> B2[main: B]
        B2 --> C2[feature: C']
        C2 --> D2[feature: D']
    end
```

```bash
# Update feature branch with latest main
git switch feature/login
git fetch origin
git rebase origin/main

# If conflict happens, edit files, then:
git add .
git rebase --continue

# Abort rebase if confused
git rebase --abort

# Clean up last 3 local commits
git rebase -i HEAD~3
```

Stash is a temporary shelf when your work is unfinished but you need to switch context.

```bash
# Save unfinished work
git stash push -m "half done login UI"

# Move elsewhere
git switch main

# Return and restore work
git switch feature/login
git stash pop

# Inspect stashes
git stash list
git stash show -p stash@{0}

# Delete stash
git stash drop stash@{0}
```

Bisect finds which commit introduced a bug.

```bash
# Start binary search
git bisect start

# Current version is broken
git bisect bad

# This older version was good
git bisect good v1.0.0

# Git checks out a middle commit
# Run your test manually, then mark result
git bisect good
git bisect bad

# End bisect and return to normal branch
git bisect reset
```

Automatic bisect is powerful when you have a test script:

```bash
# Script must exit 0 when good, non-zero when bad
git bisect start
git bisect bad
git bisect good v1.0.0
git bisect run ./run-tests.sh
git bisect reset
```

Use `.gitignore` to avoid committing generated files, secrets, virtual environments, caches, local data, and OS junk.

```gitignore
# Python cache and virtual environment
__pycache__/
*.py[cod]
.venv/
.pytest_cache/
.ruff_cache/

# Build outputs
dist/
build/
*.egg-info/

# Secrets and local env
.env
.env.local

# OS/editor files
.DS_Store
Thumbs.db
.vscode/settings.json
.idea/

# Local data
data/raw/
*.sqlite
```

Never commit secrets. If you accidentally commit `.env`, remove it from history carefully and rotate the key. Simply deleting it in a later commit is not enough because Git history still contains the old secret.

## One complete practical example

This combines local Git, GitHub, branch, commit, `.gitignore`, PR, `gh`, stash, rebase, and cleanup.

```bash
# Create GitHub repo and clone it
gh repo create git-practice-notes --public --clone
cd git-practice-notes

# Add useful files
cat > README.md <<'EOF'
# Git Practice Notes

Small project to practice Git and GitHub workflow.
EOF

cat > .gitignore <<'EOF'
# Python
__pycache__/
.venv/
.pytest_cache/

# Secrets
.env

# OS/editor
.DS_Store
.vscode/settings.json
EOF

# First commit on main
git status
git add README.md .gitignore
git commit -m "docs: add initial project notes"
git push

# Create feature branch
git switch -c feat/add-script

# Add small script
cat > hello.py <<'EOF'
name = input("Name: ")
print(f"Hello, {name}!")
EOF

# Check changes and commit
git diff
git add hello.py
git commit -m "feat(cli): add greeting script"

# Make unfinished change
echo "# TODO: add validation" >> hello.py

# Need to switch urgently, so stash
git stash push -m "wip: validation idea"

# Update main
git switch main
git pull

# Return to feature branch
git switch feat/add-script
git rebase main

# Restore unfinished work
git stash pop

# Decide to commit TODO too
git add hello.py
git commit -m "docs(cli): add validation todo"

# Push branch and open PR
git push -u origin feat/add-script
gh pr create --title "feat(cli): add greeting script" --body "Adds a small Python script for Git practice."

# After PR merge on GitHub, sync local main
git switch main
git pull

# Delete local feature branch
git branch -d feat/add-script

# See final history
git log --all --oneline --graph --decorate
```

## Important Q&A

**Q: Should I use `git pull` or `git fetch`?**
A: `git fetch` safely downloads remote changes so you can inspect them without altering your local files. `git pull` does a `fetch` and then immediately tries to `merge` those changes into your current branch. When in doubt or to avoid unexpected merge conflicts, use `git fetch` first.

**Q: Why shouldn't I use `git rebase` on `main`?**
A: Rebasing rewrites Git history. If you rebase commits that other people have already pulled, their local histories will diverge from the remote, causing massive headaches. Only rebase your own private feature branches before merging them.

**Q: What do I do if I accidentally committed a `.env` file with a secret?**
A: Do not just delete the file in a new commit. The secret will remain in the Git history. You must rewrite the Git history (e.g., using `git filter-repo` or `BFG Repo-Cleaner`) to completely remove the file, and more importantly, immediately revoke and rotate the leaked secret.

**`❓ Question:`**

A repository has:

> 1 initial commit

> Each developer creates 3 commits independently on separate branches

> There are 5 developers

> All branches are merged using no fast-forward merge

How many total commits will exist after all merges?

**Answer (Numeric):**

</summary>

---
---

**`Solution:`**

The key concept in this question is that a **no-fast-forward (`--no-ff`) merge always creates a separate merge commit**, even when Git could otherwise merge the branch without creating one. Therefore, we need to count both the commits created by the developers and the additional merge commits created when their branches are merged.

There is initially **1 commit** in the repository. Each of the 5 developers creates 3 independent commits, giving $5 \times 3 = 15$ developer commits. Since all 5 branches are merged using no-fast-forward merges, each branch merge contributes **one additional merge commit**.

Therefore, the total is the initial commit, plus 15 developer commits, plus 5 merge commits.

---
---

**Relevant concept breakdown**

A Git merge can be performed using a fast-forward merge or a no-fast-forward merge.

With a **fast-forward merge**, Git can simply move the branch pointer forward when there is no divergent history. No additional merge commit is required.

With a **no-fast-forward merge**, Git is explicitly instructed to create a merge commit. This preserves the fact that a separate branch was merged, even if a fast-forward would have otherwise been possible.

The question explicitly says:

`All branches are merged using no fast-forward merge`

Therefore, every one of the 5 developer branches contributes **one merge commit**.

| Component | Calculation | Commits |
|---|---:|---:|
| Initial repository commit | Given | `1` |
| Developer 1 | $3$ commits | `3` |
| Developer 2 | $3$ commits | `3` |
| Developer 3 | $3$ commits | `3` |
| Developer 4 | $3$ commits | `3` |
| Developer 5 | $3$ commits | `3` |
| Developer commits | $5 \times 3$ | `15` |
| No-FF merge commits | $5 \times 1$ | `5` |

---
---

🧠 **Remember**

The important exam rule is:

**A no-fast-forward merge creates a merge commit.**

So when you see:

- $N$ developers
- $C$ commits per developer
- No-fast-forward merging

the total can be calculated as: $\text{Total commits} = \text{Initial commits} + (N \times C) + N$

For this question: $1 + (5 \times 3) + 5 = \boxed{21}$

The extra `5` comes specifically from the **five no-fast-forward merge commits**.

**`❓ Question:`**

Which commands can create a new branch reference?

> `git branch new`

> `git checkout -b new`

> `git reset --hard HEAD~1`

> `git switch -c new`

> `git clone`

**`Solution:`**

A Git branch is essentially a **reference (ref) that points to a commit**. Creating a new branch therefore means creating a new branch reference that points to some commit, usually the current `HEAD`.

There are several Git commands that can create a branch. `git branch new` creates the branch reference without switching to it. `git checkout -b new` creates the branch and simultaneously switches to it. Similarly, `git switch -c new` creates a new branch and switches to it.

The other commands do not create a new branch reference. `git reset --hard HEAD~1` moves the current branch reference to another commit, while `git clone` creates a new local repository by copying an existing repository; it does not create a new branch named `new`.

---
---

**Understanding `git branch new`**

The command: `git branch new`

creates a new branch called `new`. It normally points the new branch at the commit currently referenced by `HEAD`.

So if:

~~~text
A ── B ── C
         ↑
        HEAD
~~~

then after: `git branch new`

we have:

~~~text
A ── B ── C
         ↑
       new
       HEAD
~~~

More precisely, `HEAD` still points wherever it pointed before; the new branch reference is simply created at the same commit.

Therefore, this command **does create a new branch reference**.

---
---

The command: `git checkout -b new`

combines two operations:

1. Create a new branch named `new`.
2. Switch to that newly created branch.

Conceptually:

~~~text
git checkout -b new
        │
        ├── create branch "new"
        │
        └── switch HEAD to "new"
~~~

So if `main` currently points to commit `C`, after the command:

~~~text
A ── B ── C
         ↑
        main
         ↑
        new
         ↑
        HEAD
~~~

The branch reference `new` has been created, so this is a correct answer.

---
---

**Understanding `git switch -c new`**

The newer Git command: `git switch -c new`

also creates a new branch and switches to it.

The `-c` option means **create a new branch**.

Therefore: `git switch -c new`   >>>>>  is conceptually similar to:  `git checkout -b new`

Both commands create the branch reference and move `HEAD` to it.

Thus, this command also satisfies the question.

---
---

**Why `git reset --hard HEAD~1` does not create a branch**

The command: `git reset --hard HEAD~1`

does something fundamentally different.

It moves the current branch reference backward by one commit and resets the working tree and index to that commit.

For example:

~~~text
Before:

A ── B ── C
         ↑
        main
~~~

After: `git reset --hard HEAD~1`

the branch moves to:

~~~text
A ── B ── C
     ↑
    main
~~~

No new branch reference is created. An existing branch reference is simply moved.

---
---

📊 **Option Comparison**

| Option | What it does | Creates a new branch reference? |
|---|---|---|
| `git branch new` | Creates branch `new` at the current commit without switching to it. | ✅ |
| `git checkout -b new` | Creates branch `new` and switches `HEAD` to it. | ✅ |
| `git reset --hard HEAD~1` | Moves the current branch reference to the previous commit and resets the working tree. | ❌ |
| `git switch -c new` | Creates branch `new` and switches `HEAD` to it. | ✅ |
| `git clone` | Creates a new local repository from an existing repository. | ❌ |

**`❓ Question:`**

In Git, which command shows the differences between your working directory and the last commit?

> `git status`

> `git diff`

> `git log`

> `git show`

</summary>

---
---

**`Solution:`**

The correct command is **`git diff`**.

Git maintains different states of your files. The **working directory** contains your current changes, while the **last commit** represents the most recently committed version.

`git diff` compares your current **unstaged changes** in the working directory against the **index (staging area)**. In the common case where nothing has been staged, this effectively shows what has changed since the last commit.

---
---

**`Relevant Concept Breakdown`**

Think of Git as having three important areas:

~~~text
Last Commit
     │
     ▼
Staging Area
     │
     ▼
Working Directory
~~~

Common commands:

| Command | Main purpose |
|---|---|
| `git status` | Shows which files are modified, staged, or untracked |
| `git diff` | Shows the actual unstaged line-by-line changes |
| `git log` | Shows commit history |
| `git show` | Shows information and changes associated with a particular commit |

For example, if you modify:

~~~text
hello.py
```

`git status` might tell you:

~~~text
modified: hello.py
~~~

But:

~~~bash
git diff
~~~

shows the **actual content changes**, such as:

~~~diff
- print("Hello")
+ print("Hello World")
~~~

---
---

**`Solving the Question`**

The question asks specifically for the command that **shows the differences**, not merely which files have changed.

Therefore:

~~~bash
git diff
~~~

is the appropriate command for viewing the actual differences in the working tree.

**Correct Answer:**

> **`git diff`** ✅

---

📊 **`Option Comparison`**

| Option | What it answers | Status |
|---|---|---|
| **`git status`** | "Which files have changed or been staged?" | ❌ |
| **`git diff`** | "What exactly changed in my working files?" | ✅ |
| **`git log`** | "What commits have been made?" | ❌ |
| **`git show`** | "What does a particular commit contain?" | ❌ |

---

🧠 **`Remember`**

> **`git status` → What changed?**  
> **`git diff` → What exactly changed?**  
> **`git log` → What commits exist?**  
> **`git show` → What is inside a commit?**

One useful distinction:

$$
\boxed{\texttt{git diff} = \text{Working Directory} \leftrightarrow \text{Staging Area}}
$$

To compare the working directory against the **last commit directly**, you can also use:

~~~bash
git diff HEAD
~~~

</details>

---
---

---
---

<details>

<summary>

**`❓ Question:`**

Two team members both modify the same line in `analysis.py` on different branches. When the second person tries to merge their branch, what will happen?

> Git automatically keeps the newer change

> Git creates a merge conflict that must be resolved manually by choosing which change to keep

> Git deletes both changes

> Git randomly picks one change

</summary>

---
---

**`Solution:`**

This question tests the concept of a **Git merge conflict**.

When two branches modify the **same line of the same file in different ways**, Git cannot safely determine which version should be kept. Git can automatically merge many changes, but when the changes overlap, human judgment is required.

Git therefore marks the conflicting section and pauses the merge until the developer resolves the conflict.

---
---

**`Relevant Concept Breakdown`**

Suppose the original code is:

~~~text
result = calculate_score()
~~~

Branch A changes it to:

~~~text
result = calculate_score(data)
~~~

Branch B changes the same line to:

~~~text
result = calculate_score(input_data)
~~~

When Git tries to merge them:

~~~text
                 main
                  │
          ┌───────┴───────┐
          ▼               ▼
      Branch A         Branch B
          │               │
          │ modify        │ modify
          │ same line     │ same line
          └───────┬───────┘
                  ▼
             Git Merge
                  │
                  ▼
           Merge Conflict ⚠️
                  │
                  ▼
           Manual Resolution
~~~

Git may mark the conflict in the file like:

~~~text
<<<<<<< HEAD
result = calculate_score(data)
=======
result = calculate_score(input_data)
>>>>>>> other-branch
~~~

The developer must decide what the final code should be, remove the conflict markers, and then complete the merge.

---
---

**`Solving the Question`**

The important detail is:

> **Both team members modify the same line differently.**

Because the changes overlap, Git cannot determine which change represents the intended final code.

Therefore, Git reports a **merge conflict** rather than automatically choosing one version.

The developer must manually resolve the conflicting code and then commit the resolved merge.

**Correct Answer:**

> **Git creates a merge conflict that must be resolved manually by choosing which change to keep.** ✅

---

📊 **`Option Comparison`**

| Option | Git's behavior | Status |
|---|---|---|
| **A** | Git does not decide that the newer change is automatically correct. | ❌ |
| **B** | Overlapping conflicting changes require manual resolution. | ✅ |
| **C** | Git does not simply delete both developers' changes. | ❌ |
| **D** | Git never randomly chooses a version. | ❌ |

---

🧠 **`Remember`**

> **Same lines + conflicting changes = Merge Conflict ⚠️**

Git can automatically merge **non-overlapping changes**, but when two branches make conflicting changes to the same part of a file:

$$
\boxed{\text{Git cannot decide} \rightarrow \text{Manual conflict resolution}}
$$

The developer chooses or combines the correct changes, then completes the merge.

</details>

---
---

---
---

<details>

<summary>

**`❓ Question:`**

Before pushing your changes, you want to make sure your branch has the latest updates from the main branch. Which command downloads and integrates those updates?

> `git commit main`

> `git pull origin main` (while on your feature branch)

> `git delete main`

> `git init main`

</summary>

---
---

**`Solution:`**

The correct command is **`git pull origin main`**, when you are currently on your feature branch.

`git pull` downloads changes from a remote repository and then integrates them into your current local branch. Here, `origin` refers to the remote repository, and `main` specifies the remote `main` branch from which you want to get the latest changes.

So, if you are working on `feature-login` and want the latest `main` changes:

~~~text
main
 │
 │ latest updates
 ▼
origin/main
 │
 │ git pull origin main
 ▼
feature-login
 │
 └── Your work + latest main changes
~~~

---
---

**`Relevant Concept Breakdown`**

The command:

~~~bash
git pull origin main
~~~

is essentially a combination of:

~~~bash
git fetch origin main
git merge origin/main
~~~

First, Git **downloads** the latest information from the remote `main` branch. Then it **integrates** those changes into the branch you are currently on.

For example:

~~~text
Before:

main:           A ── B ── C
                       │
feature:               └── D ── E


After git pull origin main:

main:           A ── B ── C
                       │
feature:               └── D ── E ── M
                              ▲
                              │
                     main changes integrated
~~~

If the changes conflict with your work, Git may produce a **merge conflict**, which must be resolved manually.

---
---

**`Solving the Question`**

The question asks for a command that:

1. Gets the latest changes from the remote `main` branch.
2. Integrates those changes into your current feature branch.

`git pull origin main` does exactly this.

- `git pull` → download and integrate remote changes
- `origin` → name of the remote repository
- `main` → branch to pull from

**Correct Answer:**

> **`git pull origin main` (while on your feature branch)** ✅

---

📊 **`Option Comparison`**

| Command | Purpose | Result |
|---|---|---|
| **`git commit main`** | Creates a commit; does not download `main` updates. | ❌ |
| **`git pull origin main`** | Fetches the remote `main` branch and integrates it into the current branch. | ✅ |
| **`git delete main`** | Not the command for updating a branch. | ❌ |
| **`git init main`** | Initializes a Git repository; does not download branch updates. | ❌ |

---

🧠 **`Remember`**

> **`git pull` = Fetch + Integrate**

$$
\boxed{\texttt{git\ pull\ origin\ main} \approx \texttt{git\ fetch} + \texttt{git\ merge}}
$$

And the phrase **"while on your feature branch"** matters: the changes from `main` are integrated **into the branch you currently have checked out**.

</details>

---
---

---
---

<details>

<summary>

**`❓ Question:`**

What is the main purpose of creating a pull request (PR) instead of directly merging your branch into main?

> Pull requests are faster than merging

> Pull requests allow team members to review your code, suggest changes, and approve before merging

> Pull requests automatically fix bugs

> Pull requests are only for documentation

</summary>

---
---

**`Solution:`**

The main purpose of a **Pull Request (PR)** is to introduce a **review and collaboration step** before changes are merged into the `main` branch.

Instead of immediately merging your feature branch, you open a PR so that other team members can inspect the code, discuss it, suggest improvements, and verify that it meets the project's requirements. Automated checks such as tests and linting can also run as part of the PR process.

This makes `main` more controlled and helps prevent incorrect or unfinished code from being merged directly.

---
---

**`Relevant Concept Breakdown`**

A typical collaborative workflow is:

~~~text
Feature Branch
      │
      │ Push changes
      ▼
   Pull Request
      │
      ├── Code Review 👀
      ├── Suggestions 💬
      ├── Automated Tests 🧪
      └── Approval ✅
              │
              ▼
           main
~~~

Without a PR:

~~~text
Feature Branch ───────────────► main
                 Direct merge
```

With a PR, the merge becomes a **controlled checkpoint**.

A reviewer might notice:

~~~text
"Logic looks incorrect here."
"Please add a test."
"This variable should be renamed."
~~~

The developer can make additional commits to the same PR before it is approved and merged.

---
---

**`Solving the Question`**

The question asks why a PR is preferred over directly merging a branch into `main`.

The important benefit is **code review before integration**.

A PR allows the team to:

- Review the proposed changes.
- Discuss the implementation.
- Request modifications.
- Run automated checks.
- Approve the changes before merging.

Therefore:

**Correct Answer:**

> **Pull requests allow team members to review your code, suggest changes, and approve before merging** ✅

---

📊 **`Option Comparison`**

| Option | What it actually provides | Status |
|---|---|---|
| **Pull requests are faster than merging** | A PR adds a review process; speed is not its main purpose. | ❌ |
| **Review, suggestions & approval** | Creates a controlled process for reviewing changes before they enter `main`. | ✅ |
| **Automatically fix bugs** | PRs can run automated tests, but they do not automatically fix bugs. | ❌ |
| **Only for documentation** | PRs are primarily used to review and merge code changes, not just documentation. | ❌ |

---

🧠 **`Remember`**

> **Pull Request = Review before Merge** 👀 → ✅ → 🔀

Think of a PR as a **checkpoint between your work and the main branch**:

$$
\boxed{\text{Feature} \rightarrow \text{PR} \rightarrow \text{Review} \rightarrow \text{Approval} \rightarrow \text{Merge}}
$$
